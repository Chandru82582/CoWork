```python
from sqlalchemy import create_engine



def connect_to_mysql(db_user, db_password, db_host, db_port, db_name):
    try:
            # Create a SQLAlchemy engine to connect to MySQL
            engine = create_engine(f'mysql+mysqlconnector://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}')
            connection = engine.raw_connection()
            return connection
    except Exception as e:
        print(f"An error occurred while connecting to MySQL: {e}")
        return None
```

```python
# CSV file path
CSV_FILE_PATH = 'telecom_churn.csv'

# Database credentials
DB_USER = 'root'  # Replace with your MySQL username
DB_PASSWORD = 'root'  # Replace with your MySQL password
DB_HOST = 'localhost'
DB_PORT = 3306
DB_NAME = 'telecom_churn_db'  # Replace with your desired database name

conn = connect_to_mysql(DB_USER, DB_PASSWORD, DB_HOST, DB_PORT, DB_NAME)
```

```python
# from sqlCon import connect_to_mysql

import csv




def insert_data_to_mysql(csv_file, db_user, db_password, db_host, db_port, db_name):
    """
    Reads data from a CSV file and inserts it into a MySQL table without using pandas.
    """
    try:
        connection = connect_to_mysql(db_user, db_password, db_host, db_port, db_name)
        cursor = connection.cursor()

        with open(csv_file, 'r') as f:
            reader = csv.reader(f)
            header = next(reader) # Read header row
            
            # Create table dynamically based on CSV header
            # Assuming all columns are TEXT for simplicity. You might need more sophisticated type inference.
            columns_with_types = [f"`{col}` TEXT" for col in header]
            create_table_query = f"CREATE TABLE IF NOT EXISTS telecom_data ({', '.join(columns_with_types)})"
            cursor.execute(create_table_query)
            
            # Prepare the INSERT statement
            placeholders = ', '.join(['%s'] * len(header))
            columns = ', '.join([f"`{col}`" for col in header])
            insert_query = f"INSERT INTO telecom_data ({columns}) VALUES ({placeholders})"

            # Insert data row by row
            for row in reader:
                cursor.execute(insert_query, row)

        connection.commit()
        print(f"Data from {csv_file} successfully inserted into table 'telecom_data' in database {db_name}.")

    except Exception as e:
        print(f"An error occurred: {e}")
    finally:
        if 'cursor' in locals() and cursor:
            cursor.close()
        if 'connection' in locals() and connection:
            connection.close()

```

```python
# Call the function to insert data

insert_data_to_mysql(CSV_FILE_PATH, DB_USER, DB_PASSWORD, DB_HOST, DB_PORT, DB_NAME)
```
