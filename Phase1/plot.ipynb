```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("telecom_churn.csv")
```

```python
df.head()
```

```python
df.dtypes
```

```python
print(df[['age', 'estimated_salary']].corr())
```

```python
df[['age',
          'estimated_salary',
          'calls_made',
          'sms_sent',
          'data_used',
          'churn']].corr()
```

```python
age_salary = df.groupby('age')['estimated_salary'].mean().reset_index()
```

```python

# Average salary for each age
age_salary = df.groupby('age')['estimated_salary'].mean().reset_index()

plt.figure(figsize=(8,6))
plt.scatter(age_salary['age'],
            age_salary['estimated_salary'],
            color='blue',
            s=50)

plt.xlabel('Age')
plt.ylabel('Average Estimated Salary')
plt.title('Unique Ages vs Average Estimated Salary')
plt.grid(True)
plt.show()
```

```python

plt.figure(figsize=(8,6))
sns.scatterplot(
    data=age_salary,
    x='age',
    y='estimated_salary',
    alpha=0.6
)

plt.title('Age vs Estimated Salary')
plt.show()
```
