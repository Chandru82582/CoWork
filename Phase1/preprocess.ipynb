```python
import pandas as pd
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

```python
print(df.shape)
df.dtypes
```

```python
null_summary = pd.DataFrame({
    'null_count': df.isnull().sum(),
    'null_percentage': (df.isnull().sum() / len(df)) * 100
})

print(null_summary)
```

```python
df.nunique()
```

```python
churn_count = pd.DataFrame({
    "Status": ["Churned", "Not Churned"],
    "Count": [
        df[df['Churn'] == "Yes"].shape[0],
        df[df['Churn'] == "No"].shape[0]
    ],
    "Percentage": [
        round(df[df['Churn'] == "Yes"].shape[0] / len(df) * 100, 2),
        round(df[df['Churn'] == "No"].shape[0] / len(df) * 100, 2)
    ]
})

churn_count
```
