```python
import pandas as pd
```

```python

# Create a dataframe with the order data
df = pd.DataFrame({
    'OrderID': [1001, 1002, 1003],
    'Customer': ['John', 'Mary', 'David'],
    'Product': ['Laptop', 'Mouse', 'Keyboard'],
    'Category': ['Electronics', 'Electronics', 'Electronics'],
    'Price': [55000, 500, 1200],
    'Quantity': [1, 2, 1],
    'City': ['Chennai', 'Delhi', 'Chennai'],
    'OrderDate': ['2025-01-01', '2025-01-02', '2025-01-03'],
    'Rating': [4.5, 4.2, 4.8]
})

df
```
